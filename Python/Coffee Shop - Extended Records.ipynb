## Handling Coulms


```python
#Split datetime column into date and time
df[['date','time']] = df['datetime'].str.split(' ',expand = True)
```

```python
#convert the date column to date so i can extract year , month and day from it
df['date'] = pd.to_datetime(df['date'])
df['year'] = df['date'].dt.year
df['month'] = df['date'].dt.month
df['day_of_week'] = df['date'].dt.day_name()
```

```python
#convert the date column to datetime so i can extract hour from it
df['datetime'] = pd.to_datetime(df['datetime'])
df['hour'] = df['datetime'].dt.hour
```

```python
def calss_time(hour):
    if 5 <= hour < 12:
        return 'Morning'
    elif 12 <= hour <17:
        return 'Afternoon'
    else:
        return 'Evening'
```

```python
df['time_period'] = df['hour'].apply(calss_time)
```

```python
#delet datetime coulmn
df.drop(columns='datetime' ,inplace = True)
```

```python
df['price_tier'] = pd.qcut(df['money'] ,q=3 ,labels=['Low','Medium','High'])
```
